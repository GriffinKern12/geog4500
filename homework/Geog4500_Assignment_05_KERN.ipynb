Using `ga_tracts_access.geojson` file that you uploaded to your GitHub repository, create an interactive choropleth map using `nearest_hosp_min` column (the shortest travel time to the nearst hospital).

```python
# Your code here:
import geopandas as gpd
import folium

url = "https://raw.githubusercontent.com/GriffinKern12/geog4500/main/data/ga_tracts_access.geojson"
gdf = gpd.read_file(url)
gdf = gdf.to_crs(epsg=4326)

m = folium.Map(location=[32.7, -83.3], zoom_start=7, tiles="CartoDB positron")

folium.Choropleth(
    geo_data=gdf.to_json(),
    data=gdf,
    columns=[gdf.index, "nearest_hosp_min"],
    key_on="feature.id",
    fill_color="YlOrRd",
    fill_opacity=0.7,
    line_opacity=0.2,
    legend_name="Travel time to nearest hospital (minutes)"
).add_to(m)

folium.GeoJson(
    gdf.to_json(),
    tooltip=folium.GeoJsonTooltip(
        fields=["nearest_hosp_min"],
        aliases=["Nearest hospital travel time (min):"]
    )
).add_to(m)

m
```
